# 00 — OneCore project contract

This notebook records the non-negotiable research boundary. OneCore is a single randomly initialized multimodal model. No pretrained model weights, teacher outputs, external inference APIs, or other training hosts are permitted.

All training, checkpoint lineage, evaluation, and promotion are managed by GitHub Actions. The notebook itself does not launch training.

```python
from pathlib import Path
import json

from pathlib import Path

def find_project_root(start=None):
    current = Path(start or Path.cwd()).resolve()
    for candidate in (current, *current.parents):
        if (candidate / "configs" / "phone-research.yaml").is_file() and (candidate / "src" / "onecore").is_dir():
            return candidate
    raise RuntimeError(f"Could not locate OneCore project root from {current}")

root = find_project_root()
print("OneCore project root:", root)

project = root
assert (project / "src" / "onecore").is_dir()
print('Training host: GitHub Actions only')
print('Model weights: random initialization; no imported base checkpoint')
print('Project files:', sum(1 for p in project.rglob('*') if p.is_file()))
```
