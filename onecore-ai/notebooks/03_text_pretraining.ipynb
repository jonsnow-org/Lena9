# 03 — Initial language grounding for the unified model

This is the first curriculum phase, not a text-only product or a separate language model. It establishes shared token and instruction representations before the same weights continue through image, sound, video, and action examples. Every task uses the same OneCore checkpoint.


```python
import os, sys, json
from pathlib import Path

from pathlib import Path

def find_project_root(start=None):
    current = Path(start or Path.cwd()).resolve()
    for candidate in (current, *current.parents):
        if (candidate / "configs" / "phone-research.yaml").is_file() and (candidate / "src" / "onecore").is_dir():
            return candidate
    raise RuntimeError(f"Could not locate OneCore project root from {current}")

root = find_project_root()
print("OneCore project root:", root)

sys.path.insert(0, str(root / "src"))
from onecore.train import run_training

stage = os.environ.get("ONECORE_STAGE", "text")
steps = int(os.environ.get("ONECORE_STEPS", "1"))
resume = os.environ.get("ONECORE_RESUME", "false").lower() == "true"
report = run_training(
    project_root=root,
    config_path=root / "configs" / "phone-research.yaml",
    stage=stage,
    steps=steps,
    output_dir=root / "run-state",
    resume=resume,
)
print(json.dumps(report, ensure_ascii=False, indent=2))
```
