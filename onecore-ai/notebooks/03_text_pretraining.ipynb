# 03 — Random-init smoke / text and tool pretraining

The GitHub workflow sets `ONECORE_STAGE` to `smoke` or `text`. Smoke uses generated toy inputs solely to check the pipeline and is permanently ineligible for promotion. Text training requires reviewed `data/train.jsonl` records and a newly trained tokenizer.

```python
import os, sys, json
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'onecore-ai' / 'src'))
from onecore.train import run_training

root = Path.cwd() / 'onecore-ai'
stage = os.environ.get('ONECORE_STAGE', 'smoke')
steps = int(os.environ.get('ONECORE_STEPS', '1'))
resume = os.environ.get('ONECORE_RESUME', 'false').lower() == 'true'
report = run_training(
    project_root=root,
    config_path=root / 'configs' / 'phone-research.yaml',
    stage=stage,
    steps=steps,
    output_dir=root / 'run-state',
    resume=resume,
)
print(json.dumps(report, ensure_ascii=False, indent=2))
```
